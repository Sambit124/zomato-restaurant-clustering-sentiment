## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# 1. Remove duplicate records
reviews = reviews.drop_duplicates()
metadata = metadata.drop_duplicates()

# 2. Handle missing values
reviews["Review"] = reviews["Review"].fillna("")
reviews["Metadata"] = reviews["Metadata"].fillna("Not Available")
reviews["Pictures"] = reviews["Pictures"].fillna(0)

# 3. Convert Rating into numeric format
reviews["Rating"] = pd.to_numeric(
    reviews["Rating"], errors="coerce"
)

# 4. Convert Time into datetime format
reviews["Time"] = pd.to_datetime(
    reviews["Time"], errors="coerce"
)

# 5. Clean Cost column
metadata["Cost"] = (
    metadata["Cost"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("₹", "", regex=False)
    .str.strip()
)

metadata["Cost"] = pd.to_numeric(
    metadata["Cost"], errors="coerce"
)

# 6. Merge both datasets
df = pd.merge(
    reviews,
    metadata,
    left_on="Restaurant",
    right_on="Name",
    how="left"
)

# 7. Create new features
df["Review_Length"] = df["Review"].astype(str).str.len()

df["Word_Count"] = (
    df["Review"].astype(str).str.split().str.len()
)

df["Review_Year"] = df["Time"].dt.year

df["Review_Month"] = df["Time"].dt.month

df["Has_Pictures"] = (
    pd.to_numeric(df["Pictures"], errors="coerce")
    .fillna(0) > 0
)

# 8. Display final dataset
print("Final Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nFinal Dataset:")
display(df.head())
````

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Reviews containing pictures have a different average rating from reviews without pictures.
- Null Hypothesis (H₀): There is no significant difference in average ratings between reviews with pictures and reviews without pictures.
- Alternative Hypothesis (H₁): There is a significant difference in average ratings between the two groups.
- Test: Independent two-sample Welch's t-test.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import ttest_ind

with_pictures = df.loc[
    df["Has_Pictures"] == True, "Rating"
].dropna()

without_pictures = df.loc[
    df["Has_Pictures"] == False, "Rating"
].dropna()

t_stat, p_value = ttest_ind(
    with_pictures,
    without_pictures,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: There is a significant difference.")
else:
    print("Fail to reject H0: No significant difference was established.")
````

##### Which statistical test have you done to obtain P-Value?

I performed an **Independent Two-Sample Welch’s t-test** using Python's `ttest_ind()` function from the `scipy.stats` library. This test calculates the p-value by comparing the mean ratings of two independent groups: reviews with pictures and reviews without pictures. I used `equal_var=False` because Welch’s t-test does not assume equal variances between the two groups.

##### Why did you choose the specific statistical test?

I selected the Independent Two-Sample Welch's t-test because the objective is to compare the average ratings of two independent groups: reviews containing pictures and reviews without pictures.

Welch's t-test is suitable because it does not require the two groups to have equal variances. It helps determine whether the observed difference in average ratings is statistically significant.Answer Here.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀): The distribution of restaurant-level average ratings is the same across Budget, Moderate, Expensive, and Premium categories.
Alternative Hypothesis (H₁): At least one cost category has a different distribution of restaurant-level average ratings.
Test: Kruskal–Wallis H-test for independent groups.

#### 2. Perform an appropriate statistical test.

````python
# Kruskal-Wallis test: restaurant average ratings across cost categories

restaurant_cost_test = (
    df.groupby("Restaurant")
      .agg(Average_Rating=("Rating", "mean"), Cost=("Cost", "first"))
      .dropna()
      .copy()
)

# Create four approximately equal-sized cost groups at restaurant level.
restaurant_cost_test["Cost_Category"] = pd.qcut(
    restaurant_cost_test["Cost"].rank(method="first"),
    q=4,
    labels=["Budget", "Moderate", "Expensive", "Premium"]
)

rating_groups = [
    group["Average_Rating"].dropna().values
    for _, group in restaurant_cost_test.groupby("Cost_Category", observed=False)
]

# Remove any empty groups before running the test.
rating_groups = [group for group in rating_groups if len(group) > 0]

h_stat, p_value = stats.kruskal(*rating_groups)

print("Kruskal-Wallis H-statistic:", h_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: At least one cost category differs in its rating distribution.")
else:
    print("Fail to reject H0: No statistically significant difference was established.")
````

##### Which statistical test have you done to obtain P-Value?

I performed the Kruskal–Wallis H-test using scipy.stats.kruskal(). It compares the rating distributions of three or more independent groups. The groups are restaurant-level average ratings for Budget, Moderate, Expensive, and Premium cost categories.

##### Why did you choose the specific statistical test?

I chose the Kruskal–Wallis test because the analysis compares more than two independent cost groups, and restaurant-level average ratings are not guaranteed to be normally distributed. The test is a non-parametric alternative to one-way ANOVA. A p-value below 0.05 indicates evidence that at least one group differs; it does not identify which pair differs. A post-hoc test would be needed for pairwise comparisons.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀): There is no monotonic association between review length and customer rating in the sample (Spearman ρ = 0).
Alternative Hypothesis (H₁): There is a monotonic association between review length and customer rating in the sample (Spearman ρ ≠ 0).
Test: Two-sided Spearman rank correlation.

#### 2. Perform an appropriate statistical test.

````python
# Spearman rank correlation: review length and rating

review_rating = (
    df[["Review_Length", "Rating"]]
    .apply(pd.to_numeric, errors="coerce")
    .dropna()
)

rho, p_value = stats.spearmanr(
    review_rating["Review_Length"],
    review_rating["Rating"],
    alternative="two-sided"
)

print("Spearman correlation (rho):", rho)
print("P-value:", p_value)
print("Number of reviews tested:", len(review_rating))

if p_value < 0.05:
    print("Reject H0: A statistically significant monotonic association was found.")
else:
    print("Fail to reject H0: No statistically significant monotonic association was established.")
````

##### Which statistical test have you done to obtain P-Value?

I performed a two-sided Spearman rank correlation test using scipy.stats.spearmanr(). The test returns Spearman's rho (ρ), which measures the direction and strength of a monotonic relationship, and a p-value for testing whether the population association is zero.

##### Why did you choose the specific statistical test?

I chose Spearman correlation because customer ratings are ordinal and review length may not follow a normal distribution. Spearman's test uses ranks rather than requiring a linear relationship or normally distributed variables. A statistically significant result indicates association, not causation; the value of ρ indicates whether the association is positive or negative and how strong it is.

````python
import re

# Remove URLs
df['Review'] = df['Review'].fillna('').astype(str)

def clean_urls_digits(text):
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'\b\w*\d\w*\b', '', text)
    return text

df['Review'] = df['Review'].apply(clean_urls_digits)

df[['Review']].head()
````

#### 5. Removing Stopwords & Removing White spaces

````python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = text.split()
    words = [word for word in words if word not in stop_words]
    return ' '.join(words)

df['Review'] = df['Review'].apply(remove_stopwords)

df[['Review']].head()
````

````python
# Remove extra white spaces

df['Review'] = df['Review'].str.strip()
df['Review'] = df['Review'].apply(
    lambda x: re.sub(r'\s+', ' ', x)
)

df[['Review']].head()
````

#### 6. Rephrase Text

````python
# Standardize common informal expressions

rephrase_dict = {
    "u": "you",
    "ur": "your",
    "gr8": "great",
    "luv": "love",
    "btw": "by the way"
}

def rephrase_text(text):
    words = text.split()
    words = [rephrase_dict.get(word, word) for word in words]
    return ' '.join(words)

df['Review'] = df['Review'].apply(rephrase_text)

df[['Review']].head()
````

````python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

models = {}
results = {}

def evaluate_model(name, model):
    model.fit(X_train, y_train)

    predictions = model.predict(X_test)

    results[name] = {
        'Accuracy': accuracy_score(y_test, predictions),
        'Precision': precision_score(
            y_test, predictions,
            average='weighted',
            zero_division=0
        ),
        'Recall': recall_score(
            y_test, predictions,
            average='weighted',
            zero_division=0
        ),
        'F1 Score': f1_score(
            y_test, predictions,
            average='weighted',
            zero_division=0
        )
    }

    print(name)
    print(classification_report(
        y_test, predictions,
        zero_division=0
    ))

    return model
````

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

````python
from sklearn.linear_model import LogisticRegression

model_1 = Pipeline([
    ('tfidf', TfidfVectorizer(
        max_features=3000,
        ngram_range=(1, 2)
    )),
    ('classifier', LogisticRegression(
        max_iter=1000,
        class_weight='balanced',
        random_state=42
    ))
])

models['Logistic Regression'] = evaluate_model(
    'Logistic Regression',
    model_1
)
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
from sklearn.naive_bayes import MultinomialNB

model_2 = Pipeline([
    ('tfidf', TfidfVectorizer(
        max_features=3000,
        ngram_range=(1, 2)
    )),
    ('classifier', MultinomialNB(alpha=1.0))
])

models['Multinomial Naive Bayes'] = evaluate_model(
    'Multinomial Naive Bayes',
    model_2
)
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
from sklearn.model_selection import GridSearchCV

pipeline = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('classifier', LogisticRegression(
        max_iter=1000,
        class_weight='balanced'
    ))
])

param_grid = {
    'tfidf__max_features': [2000, 3000],
    'tfidf__ngram_range': [(1, 1), (1, 2)],
    'classifier__C': [0.1, 1, 10]
}

grid_search = GridSearchCV(
    pipeline,
    param_grid,
    cv=5,
    scoring='f1_weighted',
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best CV Score:", grid_search.best_score_)

best_model = grid_search.best_estimator_

best_predictions = best_model.predict(X_test)

print(classification_report(
    y_test,
    best_predictions,
    zero_division=0
))
````

##### Which hyperparameter optimization technique have you used and why?

I have used GridSearchCV with 5-fold cross-validation for hyperparameter optimization. It systematically evaluates different combinations of hyperparameters and selects the combination that provides the best cross-validation F1-score. This helps improve model performance and reduce the risk of overfitting.
